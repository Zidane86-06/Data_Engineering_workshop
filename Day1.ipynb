```python
print("Departments in this dataset:")
print(df['department'].unique())

print('\nHow many students per department:')
print(df['department'].value_counts())
```

```python
cs_students=df[df['department']=='Computer Science']
print(f"Number of Computer Science students:{len(cs_students)}")
print("\nComputer Science Students:")
print(cs_students[['name','math_score','programming_score','attendance_percentage']])
```

```python
cs_high_attendance_students=df[(df['attendance_percentage']>90)&(df['department']=='Computer Science')]
print(f"Number of Computer Science students with attendance greater than 90%:{len(cs_high_attendance_students)}")
display(cs_high_attendance_students[['name','department','attendance_percentage']])
```

```python
df['average_score'] = df[['math_score', 'science_score', 'english_score', 'programming_score']].mean(axis=1)
display(df.head())
```

```python
top_students_per_department = df.groupby('department').apply(lambda x: x.nlargest(3, 'average_score')).reset_index(drop=True)
print("Top 3 performing students in each department:")
display(top_students_per_department[['name', 'department', 'average_score']])
```

```python
average_score_per_department = df.groupby('department')['average_score'].mean().reset_index()
print("Average score per department:")
display(average_score_per_department.sort_values(by='average_score', ascending=False))
```
