```python
import pandas as pd
```

```python
df=pd.read_csv('student_performance.csv')
```

```python
highest_math_score = df['math_score'].max()
lowest_math_score = df['math_score'].min()
average_math_score = df['math_score'].mean()

print(f"Highest Math Score: {highest_math_score}")
print(f"Lowest Math Score: {lowest_math_score}")
print(f"Average Math Score: {average_math_score:.2f}")
average_programming_score = df['programming_score'].mean()
print(f"Average Programming Score: {average_programming_score:.2f}")
average_attendance = df['attendance_percentage'].mean()
print(f"Average Attendance Percentage: {average_attendance:.2f}")
```

```python
low_attendance_students = df[df['attendance_percentage'] < 75]

if not low_attendance_students.empty:
    for index, student in low_attendance_students.iterrows():
        print(f"Name: {student['name']}, Department: {student['department']}, Attendance: {student['attendance_percentage']}%")
else:
    print("All students have attendance above 75%.")
```

```python
electronics_students_count = df[df['department'] == 'Electronics'].shape[0]
print(f"Number of students from the Electronics Department: {electronics_students_count}")
```

```python
female_students_attendance = df[df['gender'] == 'Female']['attendance_percentage'].mean()
print(f"Average attendance of female students: {female_students_attendance:.2f}%")
```

```python
lowest_programming_score_student = df.loc[df['programming_score'].idxmin()]
print(f"Student with the lowest programming score: {lowest_programming_score_student['name']} (Score: {lowest_programming_score_student['programming_score']})")
```

```python
students_above_80_math = df[df['math_score'] > 80].shape[0]
print(f"Number of students who scored above 80 in math: {students_above_80_math}")
```

```python
df['total_score'] = df['math_score'] + df['programming_score'] + df['science_score'] + df['english_score']
avg_total_score_high_attendance = df[df['attendance_percentage'] > 90]['total_score'].mean()
print(f"Average total score for students with attendance above 90%: {avg_total_score_high_attendance:.2f}")
```
